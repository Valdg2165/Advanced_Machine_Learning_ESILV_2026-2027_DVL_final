## Micro-example — uncertainty does not decide alone

A Student-t interval estimates a population mean: mean ± t(0.975, n−1) × sample standard deviation / √n. Treat complete, independently seeded episodes as sampling units; time steps within an episode are dependent. Exact coverage requires iid normal observations; skewed small samples make the approximation uncertain. It is not a prediction interval for a future episode. Compare paired per-seed cost differences on common scenarios rather than judging improvement from overlap of two separate intervals. The protected handoff currently supplies marginal intervals only, so it does not establish a paired significance claim. Release also requires service, provenance, and the comparator rule.
